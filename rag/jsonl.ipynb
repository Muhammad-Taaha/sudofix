```python
import os, gzip, json, sys
from collections import Counter

def check_first_line(filepath):
    with open(filepath, 'r') as f:
        first_line = f.readline().strip()
        try:
            data = json.loads(first_line)
            print(f"  Keys: {list(data.keys())[:5]}...")
            return True
        except:
            return False

print("Verifying security_fixes_high_confidence.jsonl")
parent_dir = os.path.dirname(os.getcwd())
data_dir = os.path.join(parent_dir, "dataset")

security_file = os.path.join(data_dir, "security_fixes_high_confidence.jsonl")

if os.path.exists(security_file):
    print(f"✅ File: {security_file}")
    check_first_line(security_file)
else:
    print(f"❌ File not found at {security_file}")

mono_dir = os.path.join(data_dir, "mono-metadata")
if os.path.exists(mono_dir):
    print(f"\n📁 Files in {mono_dir}:")
    print(os.listdir(mono_dir)[:10])
else:
    print(f"\n❌ mono-metadata directory not found")

mono_fix_file = os.path.join(data_dir, "mono_fix.py")
if os.path.exists(mono_fix_file):
    print(f"\n📄 Checking mono_fix.py...")
    with open(mono_fix_file, 'r') as f:
        print(f.read()[:500])
else:
    print(f"\n❌ mono_fix.py not found")
```
