````python
import os
import pandas as pd
import json
import sqlite3
from pathlib import Path
from unidiff import PatchSet   # optional: only if you want to parse raw patches

# ======================== CONFIGURATION =========================
DATA_PATH = "../dataset/"
DATASETS = [
    "cvedataset-patches",
    "CVEFixes.csv",
    "security_fixes_high_confidence.jsonl",
    "CVEfixes_v1.0.8.db"
]

# ======================== LOADING FUNCTIONS =====================
def load_dataset(file_name):
    """Load a single dataset (CSV, JSONL, SQLite, or directory of patches)."""
    path = os.path.join(DATA_PATH, file_name)

    if not os.path.exists(path):
        print(f"❌ Path does not exist: {path}")
        return pd.DataFrame() if not isinstance({}, dict) else {}

    # CSV
    if file_name.endswith(".csv"):
        try:
            print(f"📄 Loading CSV: {file_name}")
            return pd.read_csv(path)
        except Exception as e:
            print(f"❌ Error reading CSV {file_name}: {e}")
            return pd.DataFrame()

    # JSONL
    elif file_name.endswith(".jsonl"):
        try:
            print(f"📄 Loading JSONL: {file_name}")
            data = []
            with open(path, "r", encoding="utf-8") as f:
                for line in f:
                    data.append(json.loads(line))
            return pd.DataFrame(data)
        except Exception as e:
            print(f"❌ Error reading JSONL {file_name}: {e}")
            return pd.DataFrame()

    # SQLite database
    elif file_name.endswith(".db"):
        try:
            print(f"🗄️ Loading DB: {file_name}")
            conn = sqlite3.connect(path)
            tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
            db_data = {}
            for table in tables["name"]:
                print(f"→ Loading table: {table}")
                db_data[table] = pd.read_sql(f"SELECT * FROM {table}", conn)
            conn.close()
            return db_data
        except Exception as e:
            print(f"❌ Error reading DB {file_name}: {e}")
            return {}

    # Directory (for patch files)
    elif os.path.isdir(path):
        print(f"📂 Loading directory: {file_name}")
        rows = []
        for f in os.listdir(path):
            full_path = os.path.join(path, f)
            if f.endswith(".patch"):
                try:
                    with open(full_path, "r", encoding="utf-8", errors="ignore") as pf:
                        content = pf.read()
                    parts = Path(f).stem.split('_')
                    commit_hash = parts[-1] if len(parts) > 0 else "unknown"
                    repo = "_".join(parts[1:-1]) if len(parts) > 2 else "unknown"
                    rows.append({
                        "file_name": f,
                        "commit_hash": commit_hash,
                        "repo": repo,
                        "patch_content": content,
                        "type": "patch"
                    })
                except Exception as e:
                    print(f"❌ Error reading patch {f}: {e}")
            elif f.endswith(".csv"):
                try:
                    df_temp = pd.read_csv(full_path)
                    rows.extend(df_temp.to_dict('records'))
                except Exception as e:
                    print(f"❌ Error reading {f}: {e}")
            elif f.endswith(".jsonl"):
                try:
                    with open(full_path, "r", encoding="utf-8") as file:
                        for line in file:
                            rows.append(json.loads(line))
                except Exception as e:
                    print(f"❌ Error reading {f}: {e}")
            else:
                print(f"⚠️ Skipping unsupported file: {f}")
        if not rows:
            print("❌ No valid data found in directory!")
            return pd.DataFrame()
        return pd.DataFrame(rows)

    else:
        print(f"⚠️ Unsupported format: {file_name}")
        return pd.DataFrame()


# ======================== MAIN LOADING LOOP ======================
loaded_data = {}
for d in DATASETS:
    print("\n==============================")
    print(f"Loading: {d}")
    data = load_dataset(d)
    loaded_data[d] = data

# ======================== INSPECT LOADED DATA ====================
print("\n\n===== DATA SUMMARY =====")
for key, value in loaded_data.items():
    print(f"\n--- Dataset Key: '{key}' ---")
    if isinstance(value, pd.DataFrame):
        print(f"  ✅ Shape: {value.shape}")
        print(f"  🧭 Columns: {value.columns.tolist()}")
    elif isinstance(value, dict):
        print(f"  🗄️ SQLite DB with tables:")
        for table_name, table_df in value.items():
            print(f"     - Table: '{table_name}' (shape {table_df.shape})")
    else:
        print(f"  ⚠️ Unknown type: {type(value)}")

# ======================== EXTRACT USEFUL DATAFRAMES ==============
# 1. High‑confidence JSONL (best source)
high_conf_df = loaded_data.get('security_fixes_high_confidence.jsonl')
if isinstance(high_conf_df, pd.DataFrame):
    print(f"\n✅ High‑confidence JSONL loaded: {high_conf_df.shape}")
else:
    print("\n⚠️ High‑confidence JSONL not available")
    high_conf_df = pd.DataFrame()  # empty fallback

# 2. Raw patches (if you need to parse them)
raw_patches_df = loaded_data.get('cvedataset-patches')
if isinstance(raw_patches_df, pd.DataFrame):
    print(f"✅ Raw patches loaded: {raw_patches_df.shape}")
else:
    raw_patches_df = pd.DataFrame()

# 3. SQLite tables – SAFE access (no KeyError)
db_dict = loaded_data.get('CVEfixes_v1.0.8.db')
if isinstance(db_dict, dict):
    print(f"\n✅ SQLite DB loaded with {len(db_dict)} tables")
    # List available tables – use whatever exists
    table_names = list(db_dict.keys())
    print("   Available tables:", table_names)
    
    # Try to assign common table names – but only if they exist
    cves_table = db_dict.get('cves') or db_dict.get('CVE') or db_dict.get('cve')
    commits_table = db_dict.get('commits') or db_dict.get('Commit')
    # Add more as needed
    
    if cves_table is not None:
        print(f"   ✅ Using table '{cves_table.name if hasattr(cves_table, 'name') else '?'}' as CVE metadata (shape {cves_table.shape})")
    else:
        print("   ⚠️ No CVE metadata table found – skipping enrichment")
        cves_table = pd.DataFrame()
else:
    cves_table = pd.DataFrame()
    commits_table = pd.DataFrame()

# ======================== CLEANING HIGH‑CONFIDENCE DATAFRAME =====
def clean_high_confidence(df):
    """Basic cleaning for the JSONL dataset."""
    if df.empty:
        return df
    # Keep only rows with vulnerable and patched code
    required_cols = ['vulnerable_code', 'patched_code']
    for col in required_cols:
        if col not in df.columns:
            print(f"⚠️ Column '{col}' missing in JSONL – cannot clean")
            return df
    # Drop nulls
    df = df.dropna(subset=required_cols)
    # Remove empty strings
    df = df[df['vulnerable_code'].str.strip() != '']
    df = df[df['patched_code'].str.strip() != '']
    # Deduplicate
    df = df.drop_duplicates(subset=['vulnerable_code', 'patched_code'])
    # Optional: filter by language (if column exists)
    if 'language' in df.columns:
        allowed = ['php', 'java', 'python', 'c', 'cpp', 'javascript', 'go', 'ruby']
        df = df[df['language'].isin(allowed)]
    return df

clean_jsonl = clean_high_confidence(high_conf_df)
print(f"\n🧹 Cleaned JSONL shape: {clean_jsonl.shape}")

# ======================== ENRICH WITH CVSS/CWE (if available) ====
if not cves_table.empty and not clean_jsonl.empty:
    # Example: merge on cve_id (adjust column names as needed)
    # First, find the CVE ID column in clean_jsonl
    cve_col_jsonl = None
    for col in ['cve_id', 'CVE', 'cve']:
        if col in clean_jsonl.columns:
            cve_col_jsonl = col
            break
    # Find the CVE ID column in cves_table
    cve_col_cves = None
    for col in ['cve_id', 'CVE', 'cve']:
        if col in cves_table.columns:
            cve_col_cves = col
            break
    
    if cve_col_jsonl and cve_col_cves:
        # Merge CVSS and CWE (adjust column names)
        cvss_col = 'cvss_v3_score' if 'cvss_v3_score' in cves_table.columns else 'cvss_score'
        cwe_col = 'cwe_id' if 'cwe_id' in cves_table.columns else 'cwe'
        keep_cols = [cve_col_cves, cvss_col, cwe_col] + (['description'] if 'description' in cves_table.columns else [])
        enrich_df = cves_table[keep_cols].copy()
        clean_jsonl = clean_jsonl.merge(enrich_df, left_on=cve_col_jsonl, right_on=cve_col_cves, how='left')
        print(f"✅ Enriched with CVE metadata: {clean_jsonl.shape}")
    else:
        print("⚠️ Could not find CVE ID column for merging")
else:
    print("⚠️ Skipping enrichment – missing CVE table or JSONL")

# ======================== FINAL DATAFRAME FOR RAG ================
# This is the DataFrame you will use to build your vector index
vuln_fix_pairs = clean_jsonl

if vuln_fix_pairs.empty:
    print("\n⚠️ No usable data found. Consider parsing raw patches as fallback.")
    # Optionally, you could parse raw_patches_df here into vuln/fix pairs
else:
    print(f"\n🎯 Ready for RAG indexing: {vuln_fix_pairs.shape} rows")
    print("Columns:", vuln_fix_pairs.columns.tolist())

# ======================== OPTIONAL: EXPORT TO CSV ================
# vuln_fix_pairs.to_csv("cleaned_vuln_fix_pairs.csv", index=False)
````
